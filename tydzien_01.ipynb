## 3. Działania

Przykład z kognitywistyki: czasy reakcji (w sekundach) jednej osoby w pięciu próbach.

```python
proba1 = 0.342
proba2 = 0.401
proba3 = 0.298
proba4 = 0.455
proba5 = 0.367
proba6 = 0.512

srednia = (proba1 + proba2 + proba3 + proba4 + proba5 + proba6) / 6
print("Średni czas reakcji:", round(srednia, 3), "s")
print("W milisekundach:", round(srednia * 1000), "ms")
```

## 4. f-stringi

Najwygodniejszy sposób łączenia tekstu z liczbami.

```python
print(f"{imie} miał średni czas reakcji {srednia * 1000:.0f} ms")
```

```python
print(f"{imie} miał...")
```
