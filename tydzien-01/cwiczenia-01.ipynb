## 📌 Ściąga

| Co | Przykład | Co robi |
| --- | --- | --- |
| komentarz | `# notatka` | Python pomija wszystko po `#` |
| wypisanie | `print("Cześć")` | wyświetla tekst |
| zmienna | `wiek = 20` | zapisuje wartość pod nazwą (`=` znaczy „włóż do”) |
| tekst `str` | `"20"` | wszystko w cudzysłowie to tekst |
| liczba całkowita `int` | `20` | |
| liczba z przecinkiem `float` | `0.342` | w Pythonie przecinek dziesiętny to **kropka** |
| prawda/fałsz `bool` | `True`, `False` | |
| sprawdzenie typu | `type(wiek)` | mówi, jakiego typu jest wartość |
| działania | `+ - * /` | dodawanie, odejmowanie, mnożenie, dzielenie |
| zaokrąglanie | `round(3.14159, 2)` → `3.14` | druga liczba = ile miejsc po przecinku |
| najmniejsza / największa | `min(7, 3, 9)` → `3`, `max(7, 3, 9)` → `9` | |
| f-string | `f"Mam {wiek} lat"` | wstawia zmienną w `{ }` do tekstu |
| f-string bez miejsc po przecinku | `f"{srednia:.0f}"` | `:.1f` = 1 miejsce, `:.2f` = 2 miejsca |

**Funkcja** to nazwa, a zaraz po niej nawias z tym, co jej dajesz: `round(srednia, 1)`, `max(a, b, c)`. W środku rzeczy oddzielasz przecinkami.

⚠️ Nigdy nie nazywaj zmiennej tak jak funkcji (`print = ...`, `max = ...`), bo Python „zapomni” tę funkcję. Jeśli to się stanie, uruchom `del print` (albo inną nazwę).

## Poziom 2: co wypisze Python?

Najpierw **zgadnij** i wpisz swoją odpowiedź w komentarzu, dopiero potem uruchom. To najlepszy trening pamięci.

```python
# 2.1 Twoja odpowiedź:
x = 10
x = x + 5
print(x)
```

```python
# 2.2 Twoja odpowiedź:
a = "5"
b = "5"
print(a + b)
```

```python
# 2.3 Twoja odpowiedź:
print(7 / 2)
print(type(7 / 2))
```

```python
# 2.4 Twoja odpowiedź:
imie = "Ola"
print("Cześć, imie")
print(f"Cześć, {imie}")
```

```python
# 2.5 Twoja odpowiedź:
print(round(2.71828, 2))
print(max(4, 12, 8))
```

<details><summary>✅ Odpowiedzi do poziomu 2 (sprawdź dopiero po uruchomieniu)</summary>

- 2.1: `15`. Najpierw liczy się prawa strona (`10 + 5`), potem wynik trafia do `x`.
- 2.2: `55`. To teksty, więc `+` je skleja, a nie dodaje.
- 2.3: `3.5` i `<class 'float'>`. Dzielenie `/` zawsze daje `float`.
- 2.4: `Cześć, imie` i `Cześć, Ola`. Bez `f` i `{ }` Python nie wie, że to zmienna.
- 2.5: `2.72` i `12`.

</details>
