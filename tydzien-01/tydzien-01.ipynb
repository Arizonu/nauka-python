## 2. Zmienne i typy

Zmienna to nazwa na wartość. Najważniejsze typy: `int` (liczba całkowita), `float` (z przecinkiem), `str` (tekst), `bool` (prawda/fałsz).

```python
imie = "Kacper"
wiek = 20          # zmień na swój
czas_reakcji = 0.342  # w sekundach
student = True

print(type(imie), type(wiek), type(czas_reakcji), type(student))
```

## 3. Działania

Przykład z kognitywistyki: czasy reakcji (w sekundach) jednej osoby w pięciu próbach.

```python
proba1 = 0.342
proba2 = 0.401
proba3 = 0.298
proba4 = 0.455
proba5 = 0.367

srednia = (proba1 + proba2 + proba3 + proba4 + proba5) / 5
print("Średni czas reakcji:", round(srednia, 3), "s")
print("W milisekundach:", round(srednia * 1000), "ms")
```

## 4. f-stringi

Najwygodniejszy sposób łączenia tekstu z liczbami.

```python
print(f"{imie} miał średni czas reakcji {srednia * 1000:.0f} ms")
```
